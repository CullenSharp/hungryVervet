$$
    e^{x^2/2} = (e^{x^2})^{\frac{1}{2}} = \sqrt{e^{x^2}}
$$

$$
    \text{let} \ X = A\cos{(2\pi f t + \phi)}
$$

$$
    X_{rms} = \frac{A}{\sqrt{2}}\cos{(\phi)}
$$

```python
sr = 11025
minimum, maximum = (-3.9, 3.9)
t = np.linspace(minimum,maximum,sr)
dt = (maximum - minimum)/sr
```

```python
integrand = np.exp(-(t**2)/2) # e^(-t^2/2)
```

```python
dist = (1/np.sqrt(2*np.pi))*np.cumsum(integrand)

# normalize
# dist = (dist - dist.min())/(dist.max() - dist.min())
plt.plot(t, dist*dt)
```

```python
dist2 = (1/np.sqrt(2*np.pi))*np.trapezoid(y=integrand)
dist2
# normalize
# dist = (dist - dist.min())/(dist.max() - dist.min())
# plt.plot(t, dist2)
```

```python
import sympy as sp
```

```python
p, X, n = sp.symbols("p X n", Real=True)
```

```python
expr = (X - n*p)/(sp.sqrt(n*p*(1-p))) ; expr
```

```python
expr.simplify()
```

```python
denom = sp.sqrt(n*p*(1-p)) ; denom
```
