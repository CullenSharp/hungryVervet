````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec

def plot_mouse_samples(samples):
    # Prepare dataframe
    df = pd.DataFrame(samples)
    df["t"] = df["t"] - df["t"].min()
    df["jitter"] = np.diff(df["t"], prepend=0)

    # The window runs from (0,0) to (w,h)
    # where (0,0) is the top left corner
    # and increase in y moves up the page, not down
    df["y"] = np.flipud(df["y"])
    n = len(df)

    # Create figure and grid
    fig = plt.figure(layout="constrained")
    gs = GridSpec(
        2, 4, 
        width_ratios=[3, 10, 3, 3], 
        height_ratios=[8, 3], 
        figure=fig
    )

    # Axes setup
    ax_time_x = fig.add_subplot(gs[0])               # x vs t
    ax_3d     = fig.add_subplot(gs[1], projection="3d")  # 3D trajectory
    ax_time_y = fig.add_subplot(gs[5])               # y vs t
    ax_index  = fig.add_subplot(gs[2])               # index vs t
    ax_jitter = fig.add_subplot(gs[3])               # jitter vs index

    # Plot: x vs t
    ax_time_x.plot(df["x"], df["t"], color="r")
    ax_time_x.set(xlabel="x (px)", ylabel="t (ms)")

    # Plot: 3D trajectory
    ax_3d.plot(df["x"], df["y"], df["t"], color="orange")

    # Plot: y vs t
    ax_time_y.plot(df["t"], df["y"], color="g")
    ax_time_y.set(xlabel="t (ms)", ylabel="y (px)")

    # Plot: index vs t
    ax_index.plot(df["t"], np.arange(n), color="b")
    ax_index.tick_params(axis="y", which="both", left=False, labelleft=False)
    ax_index.set_xlabel("t (ms)")

    # Plot: jitter vs index
    ax_jitter.plot(df["jitter"], np.arange(n), color="b")
    ax_jitter.tick_params(axis="y", which="both", left=False, labelleft=False)
    ax_jitter.set_xlabel("dt")

    # Title
    fig.suptitle("Mouse Position vs Time")

    return fig, (ax_time_x, ax_3d, ax_time_y, ax_index, ax_jitter)

plot_mouse_samples(samples)
````

````python
df = pd.DataFrame(samples)
df["x"] = df["x"] - 82
xs = np.array(df["x"])
ys = np.array(df["y"])

bufferSize = 512
amplitude = np.zeros((bufferSize,bufferSize))

r = 5
i, j = np.arange(-r,r+1), np.arange(-r,r+1)
kernel = (30/r**2) * (r**2 - np.abs(np.outer(i,j))) # 5x5

h, w = amplitude.shape
# create a set of offsets from the kernel
di, dj = np.meshgrid(i,j,indexing="ij")

# flatten, turn each of these into 1dim arrays
di = di.ravel()
dj = dj.ravel()
kvals = kernel.ravel()

# apply kernel offsets to x and y
xi = (xs[:, None] + di[None, :]).ravel()
yj = (ys[:, None] + dj[None, :]).ravel()

# repeat kernal values for all pixel coords
v = np.tile(kvals, len(xs))

# assert boundary conditions
mask = (xi >= 0) & (xi < h) & (yj >= 0) & (yj < w)
xi, yj, v = xi[mask], yj[mask], v[mask]

# apply kernel
np.add.at(amplitude, (xi, yj), v)
# clamp values between 0 and 255
# return as a set of unsigned chars
np.clip(amplitude, 0, 255).astype(np.uint8)

fig, axs = plt.subplots(1,2)
axs[0].imshow(amplitude, cmap="inferno")
axs[0].set_title("Brush Effect Applied to Image")

axs[1].imshow(kernel, cmap="inferno")
axs[1].set_title("Kernel (a.k.a. brush)")
````
